```python
def print_no(n):
  if n==0:
    return

  else:
     print_no(n-1)
     print(n)

print_no(20)
```

```python
def print_num_reverse(n):
  if n==0:
    return

  else:
    print(n)
    print_num_reverse(n-1)

print_num_reverse(20)
```

```python
def sum_num(n):

  if n==0:
    return 0
  return n + sum_num(n-1)

print(sum_num(5))
```

```python
def fact(n):
  if n==0 or n==1:
    return 1

  else:
    return n*fact(n-1)
print(fact(5))
```

```python
def sqrt_natural_no(n):
  if n==1:
    return 1
  return n**2

print(sqrt_natural_no(16))
```

```python
def find_even_no(n):
  if n<=0:
    return

  find_even_no(n-1)
  if n%2==0:
    print(n)

print(find_even_no(20))
```

```python
def odd_num(n):
  if n<=0:
    return
  odd_num(n-1)
  if n%2!=0:
    print(n)

odd_num(20)
```

```python
def sum_digits(n):
  if n==0:
    return 0

  return (n % 10) + sum_digits(n  // 10)
print(sum_digits(45657676))
```

```python
def reverse_num(n):
  if n==0:
    return

  print (n%10, end="")
  reverse_num(n // 10)

reverse_num(232435)
```

```python
def count_digits(n):
  if n==0:
    return 0


  return 1 + count_digits(n // 10)

print(count_digits(3435556))

```
