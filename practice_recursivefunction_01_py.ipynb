```python
def print_no(n):
  if n==0:
    return

  else:
     print_no(n-1)
     print(n)

print_no(20)
```

```python
def print_num_reverse(n):
  if n==0:
    return

  else:
    print(n)
    print_num_reverse(n-1)

print_num_reverse(20)
```

```python
def sum_num(n):

  if n==0:
    return 0
  return n + sum_num(n-1)

print(sum_num(5))
```
