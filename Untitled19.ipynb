```python
def grade(marks):
  if 100<=marks>=80:
    return "Grade A"
  elif marks>=60:
    return "Grade B"
  elif marks>= 40:
    return "Grade C"
  else:
    return "Grade D"


print(grade(34))
```

```python
def  calculator(a, b,operator):
  if operator=="+":
    return a+b

  elif operator=="-":
    return a-b

  elif operator=="*":
    return a*b
  elif operator=="/":
    return a/b
  else:
    return "Invalid operator"
print(calculator(2,3,"+"))
print(calculator(2,3,"-"))
print(calculator(2,3,"*"))
print(calculator(4,2,"/"))
```

```python
def sum_n(n):
  sum = 0
  for i in range(n+1):
    sum+=i
  print(sum)

sum_n(5)
```

calculate factorial of the given number

```python
def fac(n):
  r = 1
  for i in range(1,n+1):
    r*=i
  print(r)

fac(0)
fac(1)
fac(5)
```

count digits of the number

```python
def count_digit(n):
  digit = 0
  while n>0:
    n//=10

    digit+=1
  print(digit)

count_digit(6)
```

```python
def reverse_no(n):
  r = 0
  while n>0:
    digit= n%10
    r = r*10+digit
    n//=10
  print(r)

reverse_no(3456)
```

```python
def print_vowel(text):
  a = 0
  vowel = "AEIOUaeiou"
  for i in text:

     if i in vowel:
      a+=1
  print(a)

print_vowel("hello")


```

```python
def reverse_string(text):
  print(text[::-1])

reverse_string("hello")
```

```python
def is_palindrome(text):
  if text == text[::-1]:
    return True

  else:
    return False

print(is_palindrome("madam"))
print(is_palindrome("tring"))

```

```python
def count_word(text):
  count=0
  for i in text:
    count+=1

  print(count)

count_word("jhiuitegfhhuo")
```

```python
def char_frequency(text):
  frequency={}
  for i in text:
    if i in frequency:
      frequency[i]+=1

    else:
      frequency[i]=1

  return frequency


print(char_frequency("sweetyicecream"))


```
