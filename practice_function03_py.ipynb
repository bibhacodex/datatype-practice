```python
def find_max(numbers):
  largest= numbers[0]
  for i in numbers:
    if i>largest:
      largest=i
  return largest

print(find_max([34,78,56,989,987,34,12]))

```

```python
def find_min(numbers:list):
  minimum =numbers[0]
  for i in numbers:
    if i<minimum:
      minimum=i
  return minimum

print(find_min([24,57,67,12,23,56,98,7,5]))
```

```python
def list_sum(items):
  sum=0
  for i in items:
    sum+=i

  return sum


print(list_sum([23,56,23]))
```

```python
def remove_duplicates(list):
  list1=[]
  for i in list:
    if i not in list1:
      list1.append(i)
  return list1



remove_duplicates([2,2,45,45,45,7880])
```

```python
def second_largest(numbers):
  largest= numbers[0]
  largest_2=numbers[0]
  for num in numbers:
    if num>largest:
      largest=num
  for i in numbers:
      if i>largest_2 and i!= largest:
         largest_2=i

  return largest_2


print(second_largest([23,567,21,45,34,46]))
```
